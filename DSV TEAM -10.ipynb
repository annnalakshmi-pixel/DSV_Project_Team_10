```python
numeric_columns = [
    "dep_delay",
    "arr_delay",
    "taxi_out",
    "taxi_in",
    "crs_elapsed_time",
    "actual_elapsed_time",
    "air_time",
    "distance"
]

outlier_results = []

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = ((df[column] < lower_limit) |
                (df[column] > upper_limit)).sum()

    outlier_results.append([
        column,
        Q1,
        Q3,
        IQR,
        lower_limit,
        upper_limit,
        outliers
    ])

outlier_table = pd.DataFrame(
    outlier_results,
    columns=[
        "Column",
        "Q1",
        "Q3",
        "IQR",
        "Lower Limit",
        "Upper Limit",
        "Outlier Count"
    ]
)

outlier_table
```

```python
import numpy as np
import pandas as pd

zscore_columns = [
    "dep_delay",
    "arr_delay",
    "taxi_out",
    "taxi_in",
    "crs_elapsed_time",
    "actual_elapsed_time",
    "air_time",
    "distance"
]

zscore_results = []

for column in zscore_columns:
    mean = df[column].mean()
    std = df[column].std()

    z_scores = ((df[column] - mean) / std).abs()

    outlier_count = (z_scores > 3).sum()

    zscore_results.append([
        column,
        mean,
        std,
        outlier_count
    ])

zscore_table = pd.DataFrame(
    zscore_results,
    columns=[
        "Column",
        "Mean",
        "Standard Deviation",
        "Z-Score Outlier Count"
    ]
)

zscore_table
```
